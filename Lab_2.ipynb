Greatest among three numbers

```python
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
c=int(input("Enter third number" ))
if(a>b and a>c):
  print(f"{a} is greatest")
elif(b>a and b>c):
  print(f"{b} is greatest")
else:
  print(f"{c} is greatest")
```

Program on "match" which alternate to "switch" in C language

```python
ch=int(input("Enter your choice:"))
match ch:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
```
