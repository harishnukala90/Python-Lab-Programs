```python
n=int(input())
count=0
for i in range(1,n):
  if n%i == 0:
    count+=1
if(count!=1):
  print("not prime")
else:
  print("prime")
```

```python
n=int(input("Enter starting number: "))
l=int(input("Enter last number: "))
for i in range(n,l+1):
  for j in range(2,i):
    if i%j ==0:
      break
  else:
    print(i,end=" ")
```
