```python
def palindrome(a):
  d = a
  c = 0
  while(a!=0):
    b = a%10
    c = c*10+b
    a = a//10
  if(d==c):
    print("The number is a palindrome")
  else:
    print("Not a palindrome")
palindrome(1991)

```

```python
def reverse(name):
  rev = ""

  for i in range(len(name)-1,-1,-1):
    rev = rev+name[i]
  return rev
print(reverse("hello"))
```

```python
def ADD(*num):
  print(num)
  d = sum(num)
  return d
print(ADD(1,2,3,4,5,6,7,8,9))
```

```python
def multiple_values(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_count,res_avg = multiple_values(1,2,3,4,5,6,7,8,9)
print(f"Sum is {res_sum},   Count is {res_count}, AVG is {res_avg}")
```
